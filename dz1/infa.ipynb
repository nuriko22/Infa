```python
# PetalLengthCm +- линейно зависит от PetalWidthCm
```

```python
def MNK(x, y):
    n = len(x)
    x_mean = np.mean(x)
    y_mean = np.mean(y)
    xy_mean = np.mean(x * y)
    x2_mean = np.mean(x * x)

    k = (xy_mean - x_mean * y_mean) / (x2_mean - x_mean**2)
    b = y_mean - k * x_mean

    return k, b
```

```python
k, b = MNK(data['PetalLengthCm'], data['PetalWidthCm'])
print(f"k = {k}, b = {b}")
x = [1, 7]
y = [k+b, k*7+b]
plt.plot(x,y, color='r', linewidth=2, label=f"k = {k}, b = {b}")
plt.scatter(data['PetalLengthCm'], data['PetalWidthCm'])
plt.ylabel('PetalLengthCm')
plt.xlabel('PetalWidthCm')
plt.grid()
plt.show()
```

```python
k, b = MNK(data['SepalLengthCm'], data['PetalLengthCm'] ** 2.5)
print(f"k = {k}, b = {b}")
x = [4.25, 8]
y = [k*x[0]+b, k*x[1]+b]
plt.plot(x,y, color='r', linewidth=2, label=f"k = {k}, b = {b}")
plt.scatter(data['SepalLengthCm'], data['PetalLengthCm'] ** 2.5)
plt.ylabel('PetalLengthCm')
plt.xlabel('PetalLengthCm^2.5')
plt.grid()
plt.show()
```

```python
N = 1000
movement = np.random.choice([1, -1], size=N)
X = np.cumsum(movement)
```

```python
X
```

```python
plt.figure(figsize=(10, 7))
plt.scatter(np.arange(1, N + 1), X)
plt.ylabel('X')
plt.xlabel('N')
plt.title('Траектория одной частицы (1000 шагов)')
plt.grid()
plt.show()
```

```python
N = 1000
n_chastic = 1000
movement = np.random.choice([1, -1], size=(n_chastic, N))
X_final = movement.sum(axis=1)
movement
```

```python
X_final
```

```python
plt.hist(X_final, bins=100)
plt.title('Распределение конечных положений 1000 частиц')
plt.xlabel('X')
plt.show()

print(f"Среднее: {X_final.mean():.2f}")
print(f"Std:     {X_final.std():.2f}   (√1000 = {np.sqrt(N):.2f})")
```

```python
# Позиции каждой частицы на каждом шаге: shape (1000, 1000)
positions = np.cumsum(movement, axis=1)
# Для каждого N считаем sqrt(<x^2>) по всем частицам
kor = np.sqrt((positions ** 2).mean(axis=0))
# Теория:
N = np.arange(1, N + 1)
theory = np.sqrt(N)

plt.figure(figsize=(10, 7))
plt.plot(N, kor, label='эксперимент: √⟨x²⟩')
plt.plot(N, theory, '--', color='r', linewidth=2, label='теория: √N')
plt.xlabel('N - количество шагов')
plt.ylabel('√⟨x²⟩')
plt.legend()
plt.grid()
plt.show()
```
